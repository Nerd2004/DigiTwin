```python
# ================================================================
# Step 2: Import Libraries and Load the XTTS Model (with PyTorch Fix)
# ================================================================
print("\nImporting libraries and loading model...")
import os
import time
import torch

# --- PyTorch >= 2.6 Safe Loading Fix ---
import torch.serialization
try:
    from TTS.tts.configs.xtts_config import XttsConfig
    from TTS.tts.models.xtts import XttsAudioConfig
    from TTS.config.shared_configs import BaseDatasetConfig
    from TTS.tts.models.xtts import XttsArgs
    safe_globals_list = [XttsConfig, XttsAudioConfig, BaseDatasetConfig, XttsArgs]
    torch.serialization.add_safe_globals(safe_globals_list)
    print(f"Attempted to add {len(safe_globals_list)} classes to safe globals for PyTorch loading.")
except ImportError as ie:
    print(f"Warning: Could not import one or more required TTS classes: {ie}. Loading might fail.")
except AttributeError:
    print("Warning: torch.serialization or add_safe_globals not found (PyTorch version issue?). Loading might fail.")
# --- End PyTorch Fix ---

from TTS.api import TTS
from IPython.display import Audio, display

# Check for GPU
device = "cuda" if torch.cuda.is_available() else "cpu"
if device == "cpu":
    print("WARNING: CUDA not available, running on CPU. This will be VERY SLOW.")
else:
    print(f"PyTorch version: {torch.__version__}")
    print(f"Using device: {device}")

# --- Load the TTS Model ---
audio_model_name = "tts_models/multilingual/multi-dataset/xtts_v2"
print(f"\nLoading XTTS model: {audio_model_name}")
print("This might take several minutes for the first download...")

load_start_time = time.time()
tts_model = None # Initialize tts_model
try:
    tts_model = TTS(audio_model_name).to(device)
    load_end_time = time.time()
    print(f"Model loaded successfully in {load_end_time - load_start_time:.2f} seconds.")
except Exception as e:
    print(f"\n--- FATAL ERROR LOADING MODEL ---")
    print(f"{e}")
    print("Cannot proceed without a loaded model. Please check previous errors, restart runtime, and try again.")
    # Optional: raise e
```

```python
from transformers import AutoModelForCausalLM, AutoTokenizer

# Specify the base model name
text_model_name = "TheBloke/Mistral-7B-Instruct-v0.2-GPTQ"

# Load the base model
text_model = AutoModelForCausalLM.from_pretrained(text_model_name,
                                             device_map="auto",
                                             trust_remote_code=False,
                                             revision="main")

# Load the tokenizer
tokenizer = AutoTokenizer.from_pretrained(text_model_name, use_fast=True)

# Set the padding token to be the same as the eos_token
tokenizer.pad_token = tokenizer.eos_token
```

```python
# ================================================================
# Step 3: Define the Voice Cloning Function
# ================================================================
import time
def clone_voice(reference_audio_path, text_to_speak, language="en", output_path="cloned_speech.wav"):
    """
    Uses the pre-loaded tts_model to generate speech directly.

    Args:
        reference_audio_path (str): Path to the input reference audio file.
        text_to_speak (str): The text to be synthesized.
        language (str, optional): Language code for synthesis. Defaults to "en".
        output_path (str, optional): Path to save the generated WAV file. Defaults to "cloned_speech.wav".

    Returns:
        str: The path to the generated audio file if successful, None otherwise.
    """
    global tts_model # Access the globally loaded model


    if tts_model is None:
        print("ERROR: TTS Model not loaded. Please ensure Step 2 ran successfully.")
        return None
    if not os.path.exists(reference_audio_path):
        print(f"ERROR: Reference audio not found at: {reference_audio_path}")
        return None
    if not text_to_speak or not isinstance(text_to_speak, str) or text_to_speak.strip() == "":
         print("ERROR: Invalid text provided. Please provide a non-empty string.")
         return None

    print("\n--- Starting Voice Cloning ---")
    print(f" > Reference: {reference_audio_path}")
    print(f" > Text: \"{text_to_speak[:100]}...\"") # Log snippet
    print(f" > Language: {language}")
    print(f" > Output: {output_path}")

    gen_start_time = time.time()
    try:
        tts_model.tts_to_file(
            text=text_to_speak,
            speaker_wav=reference_audio_path,
            language=language,
            file_path=output_path
        )
        gen_end_time = time.time()
        generation_time = gen_end_time - gen_start_time

        if os.path.exists(output_path):
            print(f"\nSuccess! Generated in {generation_time:.2f} seconds.")
            return output_path
        else:
            print("ERROR: TTS generation finished but output file not found.")
            return None

    except RuntimeError as e:
        print(f"\nRUNTIME ERROR during generation: {e}")
        if "CUDA out of memory" in str(e):
            print("--> Try using shorter text or restarting the Colab runtime to free GPU memory.")
        return None
    except Exception as e:
        print(f"\nUNEXPECTED ERROR during generation: {e}")
        return None
```

```python
import gradio as gr
import fitz  # PyMuPDF
from PIL import Image
import shutil
import os
import concurrent.futures
import re
import requests
import time
from pydub import AudioSegment
query_engine_global = None  # Global variable to store the query engine


# Global variable to store instruction
instruction_text = ""


def handle_upload(doc_path, audio_path, image_path,instruction_input):

    global instruction_text
    instruction_text = instruction_input  # Store instruction globally

    doc_text = ""
    saved_doc_path, saved_audio_path, saved_image_path = None, None, None

    upload_dir = "/content/uploaded"
    os.makedirs(upload_dir, exist_ok=True)


    # Clean up existing files
    for f in os.listdir(upload_dir):
        file_path = os.path.join(upload_dir, f)
        if os.path.isfile(file_path):
            os.remove(file_path)



    if doc_path:
          doc_ext = os.path.splitext(doc_path)[1]  # Get file extension
          saved_doc_path = os.path.join(upload_dir, f"uploaded_document{doc_ext}")
          shutil.copy(doc_path, saved_doc_path)  # Save the uploaded document

          if doc_ext == ".pdf":
              # Read the PDF and convert to text
              doc = fitz.open(saved_doc_path)
              doc_text = ""  # Initialize empty string to store text
              for page in doc:
                  doc_text += page.get_text()  # Extract text from each page
              doc.close()

              # Save the extracted text into a .txt file
              text_doc_path = os.path.join(upload_dir, "uploaded_document.txt")
              with open(text_doc_path, 'w', encoding='utf-8') as f:
                  f.write(doc_text)  # Write the extracted text into the .txt file

              # Set the new document path to the text document
              saved_doc_path = text_doc_path  # Use the text document for uploading

          else:
              # If the document is already a text file (or non-PDF), just read it
              with open(saved_doc_path, 'r', encoding='utf-8', errors='ignore') as f:
                  doc_text = f.read()

    if audio_path:
        audio = AudioSegment.from_file(audio_path)
        saved_audio_path = os.path.join(upload_dir, "uploaded_audio.wav")
        audio.export(saved_audio_path, format="wav")

    if image_path:
        image = Image.open(image_path).convert("RGB")  # Convert to RGB to ensure PNG compatibility
        saved_image_path = os.path.join(upload_dir, "uploaded_image.png")
        image.save(saved_image_path, format="PNG")

    return doc_text[:1000], saved_audio_path, saved_image_path, instruction_input


def create_clone_with_timeout(doc_text, audio_path, image_path, timeout_duration=300):
    def task():
        return create_clone(doc_text, audio_path, image_path)

    with concurrent.futures.ThreadPoolExecutor(max_workers=1) as executor:
        future = executor.submit(task)

        try:
            result = future.result(timeout=timeout_duration)
            return "✅ Clone created successfully!"
        except concurrent.futures.TimeoutError:
            print("⏰ Clone creation timed out after", timeout_duration, "seconds")
            return "❌ Clone creation timed out. Please try again with smaller files or simpler queries."


def create_clone(doc_text, audio_path, image_path):
    global query_engine_global

    try:
        print("🚀 Starting clone creation...")
        print(f"📝 Text preview:\n{doc_text[:500]}")
        print(f"🔊 Audio path: {audio_path}")
        print(f"🖼️ Image path: {image_path}")

        documents_dir = "/content/uploaded"
        txt_files = [f for f in os.listdir(documents_dir) if f.endswith('.txt')]

        documents = SimpleDirectoryReader(
            input_files=[os.path.join(documents_dir, f) for f in txt_files]
        ).load_data(show_progress=True)

        index = VectorStoreIndex.from_documents(documents)

        retriever = VectorIndexRetriever(index=index, similarity_top_k=5)
        query_engine = RetrieverQueryEngine(
            retriever=retriever,
            node_postprocessors=[SimilarityPostprocessor(similarity_cutoff=0.5)],
        )

        query_engine_global = query_engine
        print("✅ Clone created!")
        return "✅ Clone created!"

    except Exception as e:
        print("❌ Error during clone creation:", str(e))
        return f"❌ Error occurred: {str(e)}"


def ask_clone_with_timeout(query, timeout_duration=1000):
    def task():
        return ask_clone(query)

    with concurrent.futures.ThreadPoolExecutor(max_workers=1) as executor:
        future = executor.submit(task)
        try:
            return future.result(timeout=timeout_duration)
        except concurrent.futures.TimeoutError:
            return "❌ Response timed out. Please try again with a shorter question or less load."


def ask_clone(query):
    # import time
    # from IPython.display import Video, display
    global query_engine_global

    if query_engine_global is None:
        return gr.update(visible=False), "⚠️ Please create a clone first."

    try:
        # 🔄 Delete old output.mp4 if it exists
        output_video_path = "/content/output_video.mp4"
        if os.path.exists(output_video_path):
            try:
                os.remove(output_video_path)
                print("🗑️ Old output.mp4 deleted.")
            except Exception as e:
                print(f"⚠️ Failed to delete old output.mp4: {e}")

        # =============================
        # Step 1: Get LLM Answer
        # =============================
        response = query_engine_global.query(query)
        context = "Context:\n"
        # for i in range(5):
        #     context += response.source_nodes[i].text + "\n\n"

        for node in response.source_nodes[:5]:  # Only takes up to 5, no error
            context += node.text + "\n\n"

        comment = query


        prompt_template_w_context = lambda context, comment: f"""{instruction_text}

        {context}
        Please respond to the following comment as that person in only 100 words. Use the context above to answer as first person and say you don't have any answer if relevant information is not available in context and answer as first person necessarily.

        {comment}
        [/INST]"""



        prompt = f"{prompt_template_w_context(context, comment)}\nResponse:"

        inputs = tokenizer(prompt, return_tensors="pt")
        # Tokenize the input
        # inputs = tokenizer(prompt, return_tensors="pt", padding=True, truncation=True, max_length=1024)
        # outputs = text_model.generate(
        #     input_ids=inputs["input_ids"].to("cuda"),
        #     max_new_tokens=200,
        #     eos_token_id=tokenizer.eos_token_id
        # )
        # generated_text = tokenizer.decode(outputs[0], skip_special_tokens=True)

        # Use torch.no_grad() to disable gradient calculations
        with torch.no_grad():
            # Move input tensors to GPU (if available)
            input_ids = inputs["input_ids"].to("cuda")

            # Generate text
            # outputs = text_model.generate(
            #     input_ids=input_ids,
            #     max_new_tokens=150,  # Reduce max tokens to speed up inference
            #     eos_token_id=tokenizer.eos_token_id,
            #     num_return_sequences=1,  # You can adjust this if you want multiple responses
            # )
            outputs = text_model.generate(
                  input_ids=input_ids,
                  max_new_tokens=150,  # Allowing more tokens for better response completeness
                  eos_token_id=tokenizer.eos_token_id,  # Ensuring proper stop token handling
                  num_return_sequences=1,
                  temperature=0.7,  # Control creativity, avoid randomness in abrupt endings
                  top_p=0.9,  # Nucleus sampling for smoother responses
                  pad_token_id=tokenizer.eos_token_id  # Use eos token if padding is needed
              )


            # Decode generated text
            generated_text = tokenizer.decode(outputs[0], skip_special_tokens=True)

        match = re.search(r"Response:\s*(.*?)(?:\n*\(Note:.*|$)", generated_text, re.DOTALL)
        answer = match.group(1).strip() if match else generated_text.strip()

        print("Response:",answer)
        # =============================
        # Step 2: Clone Voice to Shared Path
        # =============================
        shared_dir = "/content/generated"
        os.makedirs(shared_dir, exist_ok=True)

        reference_audio_path = "/content/uploaded/uploaded_audio.wav"
        generated_audio_path = os.path.join(shared_dir, "speech.wav")

        if tts_model is not None:
            generated_file_path = clone_voice(
                reference_audio_path=reference_audio_path,
                text_to_speak=answer,
                language="en",
                output_path=generated_audio_path
            )

            if generated_file_path and os.path.exists(generated_file_path):
                print(f"✅ Cloned speech saved to {generated_file_path}")
            else:
                print("❌ Failed to generate speech.")
                return "❌ Failed to generate speech."
        else:
            print("❌ TTS model not loaded.")
            return "❌ TTS model not loaded."

        # Your actual ngrok URL from Colab
        API_URL = "https://d9f2-34-27-183-70.ngrok-free.app"

        # Local paths to your files
        image_path = "/content/uploaded/uploaded_image.png"
        audio_path = "/content/generated/speech.wav"

        # Headers for ngrok
        headers = {"ngrok-skip-browser-warning": "true"}

        # Upload files and start processing
        with open(image_path, 'rb') as img, open(audio_path, 'rb') as aud:
            response = requests.post(
                f"{API_URL}/generate",
                files={'image': img, 'audio': aud},
                headers=headers
            )

        job_id = response.json()['job_id']
        print(f"Job started with ID: {job_id}")

        # Check status until complete
        while True:
            status = requests.get(f"{API_URL}/status/{job_id}", headers=headers).json()
            print(f"Status: {status['status']}")

            if status['status'] == 'completed':
                # Download the video
                print("Downloading video...")
                video = requests.get(f"{API_URL}/download/{job_id}", headers=headers, stream=True)
                with open("output_video.mp4", 'wb') as f:
                    for chunk in video.iter_content(chunk_size=8192):
                        f.write(chunk)
                print("Video downloaded as output_video.mp4")
                break
            elif status['status'] == 'failed':
                print(f"Failed: {status.get('error', 'Unknown error')}")
                break

            time.sleep(5)  # Wait 5 seconds before checking again

        # Create a new unique temp path
        timestamp = int(time.time())
        temp_gradio_path = f"/tmp/output_{timestamp}.mp4"

        # Copy the video to the temp path
        shutil.copy(output_video_path, temp_gradio_path)

        # ✅ Show and autoplay video
        return gr.update(value=temp_gradio_path, visible=True, autoplay=True), ""



    except Exception as e:
        return f"❌ Error: {str(e)}"

# Gradio UI Setup
with gr.Blocks(title="DigiTwin") as demo:
    gr.Markdown("<h1 style='text-align: center;'>🧠 DigiTwin</h1>", elem_id="title")
    description_text = gr.Markdown("### Create your own Digital Twin by uploading your files below.")

    with gr.Row() as upload_row:
        doc_input = gr.File(label="📄 Upload PDF or Text File", file_types=[".pdf", ".txt"])
        audio_input = gr.Audio(label="🎵 Upload Audio File", type="filepath")
        image_input = gr.File(label="🖼️ Upload Image File", file_types=[".png", ".jpg", ".jpeg"])
        instruction_input = gr.Textbox(label="✍️ Instruction for Your Twin", placeholder="e.g. Speak like Sherlock Holmes...")

    submit_btn = gr.Button("🔍 Preview Uploads", visible=True)

    with gr.Row(visible=True) as preview_row:
        doc_preview = gr.Textbox(label="📄 Document Text Preview (First 1000 chars)", lines=6)
        audio_preview = gr.Textbox(label="🎵 Audio File Path")
        image_preview = gr.Image(label="🖼️ Uploaded Image Preview")
        instruction_preview = gr.Textbox(label="📝 Instruction (Saved)", interactive=False)


    clone_btn = gr.Button("✨ Create Clone", visible=True)
    clone_status = gr.Textbox(label="Clone Status", lines=2)

    gr.Markdown("### 💬 Ask your Clone")

    with gr.Row():
        with gr.Column(scale=1):
            query_input = gr.Textbox(label="Ask a Question")
            ask_btn = gr.Button("🧠 Ask")

        with gr.Column(scale=1):
            query_output = gr.Video(label="Response Video", format="mp4", visible=True)

    # Upload preview
    submit_btn.click(
        fn=handle_upload,
        inputs=[doc_input, audio_input, image_input, instruction_input],
        outputs=[doc_preview, audio_preview, image_preview, instruction_preview]
    )

    # Create clone and hide all the upload-related UI, and also hide the clone status after creation
    clone_btn.click(
        fn=create_clone_with_timeout,
        inputs=[doc_preview, audio_preview, image_preview],
        outputs=[clone_status]
    ).then(
        fn=lambda: (
            gr.update(visible=False),  # upload_row
            gr.update(visible=False),  # preview_row
            gr.update(visible=False),  # clone_btn
            gr.update(visible=False),  # submit_btn
            gr.update(visible=False),  # description_text
            gr.update(visible=False)   # clone_status (hide after creation)
        ),
        inputs=[],
        outputs=[upload_row, preview_row, clone_btn, submit_btn, description_text, clone_status]
    )

    # Ask with typed input
    ask_btn.click(
        fn=lambda x: "🕐 Generating response...",
        inputs=[query_input],
        outputs=[clone_status]
    ).then(
        fn=ask_clone_with_timeout,
        inputs=[query_input],
        outputs=[query_output, clone_status]
    )

demo.launch(share=True, debug=True, max_threads=1)



```
